```python
import os
import sys

project_root = os.path.abspath("")
if project_root not in sys.path:
    sys.path.insert(0, project_root)

from llm import get_llm
from graph import build_graph

app = build_graph(get_llm())

transcript = """
John: We need to improve the website performance.
Sarah: Yes, page load time is too slow.
David: I will optimize the database queries this week.
Sarah: I will redesign the homepage layout.
John: Let's try to finish these tasks before Friday.
"""

result = app.invoke({"transcript": transcript})
print(result["final_report"])

# Show which node ran, in order
for step in app.stream({"transcript": transcript}, stream_mode="updates"):
    print("Ran node:", list(step.keys())[0])

# Draw the workflow
print(app.get_graph().draw_mermaid())
```
